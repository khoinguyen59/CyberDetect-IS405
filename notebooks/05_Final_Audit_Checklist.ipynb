```python
from google.colab import drive
drive.mount('/content/drive')

import os
PROJECT_PATH = '/content/drive/MyDrive/Nhom28_CyberDetect_MLP_Final'
os.makedirs(PROJECT_PATH, exist_ok=True)
%cd {PROJECT_PATH}
print('PROJECT_PATH =', PROJECT_PATH)

import time
from contextlib import contextmanager

@contextmanager
def step(name):
    t0 = time.time()
    print(f"\n[START] {name}", flush=True)
    try:
        yield
    finally:
        print(f"[DONE] {name} - {time.time() - t0:.1f}s", flush=True)
```

```python

import os, pandas as pd
required = [
 'data/colab_processed/X_train_top30.parquet',
 'data/colab_processed/X_test_top30.parquet',
 'models/colab_cyberdetect_mlp.h5',
 'models/colab_preprocessing_metadata.json',
 'results/colab_table3_metrics.csv',
 'results/colab_ablation_metrics.csv',
 'results/xai/colab_shap_summary.png',
 'results/xai/colab_integrated_gradients.png',
 'bigdata/docker-compose.yml',
 'bigdata/flume/flume.conf',
 'bigdata/output/metadata.json',
]
with step('Check final artifacts'):
    rows=[]
    for r in required:
        p=f'{PROJECT_PATH}/{r}'; rows.append({'artifact':r,'status':'OK' if os.path.exists(p) else 'MISSING','size_bytes':os.path.getsize(p) if os.path.exists(p) and os.path.isfile(p) else ''})
df=pd.DataFrame(rows); display(df); df.to_csv(f'{PROJECT_PATH}/results/final_audit_checklist.csv',index=False)
for csv in ['results/colab_table3_metrics.csv','results/colab_ablation_metrics.csv','results/spark_output_table3_metrics.csv','results/spark_output_ablation_metrics.csv']:
    p=f'{PROJECT_PATH}/{csv}'
    if os.path.exists(p):
        print('\n'+csv); display(pd.read_csv(p))
```
